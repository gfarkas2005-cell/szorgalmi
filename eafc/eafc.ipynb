```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("EAFC26.csv")

# 1. A 3 legnagyobb sebességű (PAC) játékos
top3 = df.nlargest(3, "PAC")[["Name", "PAC", "Team"]]
print("1. Top 3 PAC:\n", top3.to_string(index=False))
plt.figure(figsize=(7, 4))
plt.bar(top3["Name"], top3["PAC"], color="tab:blue")
plt.ylim(90, 100)
plt.title("1. A 3 leggyorsabb játékos (PAC)")
plt.ylabel("PAC")
plt.tight_layout()
plt.savefig("1_top3_pac.png", dpi=120)
plt.close()

# 2. Átlagos OVR nemenként
nemek = df.groupby("GENDER")["OVR"].mean()
print("\n2. Átlagos OVR:\n", nemek.round(2))
plt.figure(figsize=(5, 4))
plt.bar(["Férfi (M)", "Női (F)"], [nemek["M"], nemek["F"]], color=["tab:blue", "tab:red"])
plt.title("2. Átlagos OVR nemenként")
plt.ylabel("Átlag OVR")
plt.tight_layout()
plt.savefig("2_nemek_ovr.png", dpi=120)
plt.close()

# 3. Premier League játékosok, akiknek SHO >= 85
pl = df[df["League"] == "Premier League"]
db = (pl["SHO"] >= 85).sum()
print(f"\n3. Premier League, SHO >= 85: {db} játékos")
plt.figure(figsize=(5, 4))
plt.bar(["SHO < 85", "SHO ≥ 85"], [len(pl) - db, db], color=["lightgray", "tab:green"])
plt.title("3. Premier League játékosok lövése (SHO)")
plt.ylabel("Játékosok száma")
plt.tight_layout()
plt.savefig("3_premier_league_sho.png", dpi=120)
plt.close()

# 4. Korreláció PAC és DRI között
korr = df["PAC"].corr(df["DRI"])
print(f"\n4. PAC–DRI korreláció: {korr:.3f}")
plt.figure(figsize=(6, 5))
plt.scatter(df["PAC"], df["DRI"], s=5, alpha=0.3)
plt.title(f"4. Sebesség és cselezés (r = {korr:.2f})")
plt.xlabel("PAC")
plt.ylabel("DRI")
plt.tight_layout()
plt.savefig("4_pac_dri.png", dpi=120)
plt.close()

# 5. Top 5 csapat átlagos OVR-rel és az átlagos DEF
csapat = df.groupby("Team")[["OVR", "DEF"]].mean()
top5 = csapat.nlargest(5, "OVR")
print("\n5. Top 5 csapat:\n", top5.round(2))
print("Teljes adathalmaz átlag DEF:", round(df["DEF"].mean(), 2))
x = range(len(top5))
plt.figure(figsize=(8, 4.5))
plt.bar([i - 0.2 for i in x], top5["OVR"], width=0.4, label="Átlag OVR")
plt.bar([i + 0.2 for i in x], top5["DEF"], width=0.4, label="Átlag DEF")
plt.axhline(df["DEF"].mean(), color="gray", linestyle="--", label="Összes játékos átlag DEF")
plt.xticks(list(x), top5.index, rotation=15)
plt.title("5. Top 5 csapat: OVR és DEF")
plt.legend()
plt.tight_layout()
plt.savefig("5_top5_csapat.png", dpi=120)
plt.close()
```
